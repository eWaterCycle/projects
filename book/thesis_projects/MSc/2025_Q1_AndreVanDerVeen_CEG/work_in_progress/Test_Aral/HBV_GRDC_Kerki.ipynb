## HBV

```python
model = ewatercycle.models.HBV(forcing=ERA5_forcing)
```

```python
# parameter_set =  [
#     7.085,  # Imax
#     0.837,  # Ce
#     76.373, # Sumax
#     1.112,  # Beta
#     0.245,  # Pmax
#     7.801,  # Tlag
#     0.096,  # Kf
#     0.003,  # Ks
#     0.226   # FM
# ]

parameter_set=[1.45197413e+00, 5.32414158e-01, 5.10051426e+02, 5.13793028e-01,
 1.21618379e-01, 2.70619668e+00, 1.31961887e-02, 5.94101978e-03,
 3.14826427e-01,]

initial_conditions = np.array([
    0,      # Si
    100,    # Su
    0,      # Sf
    5,      # Ss
    0       # Sp
])
```

```python
config_file, _ = model.setup(parameters=parameter_set, initial_storage=initial_conditions)
```

```python
model.initialize(config_file)
```

```python
Q_m = []
time = []
while model.time < model.end_time:
    model.update()
    Q_m.append(model.get_value("Q")[0])
    time.append(pd.Timestamp(model.time_as_datetime))
```

```python
# Finalize model (shuts down container, frees memory)
model.finalize()
```

```python
model_output = pd.Series(data=Q_m, name="Modelled_discharge", index=time)
```

```python
model_output.plot(title="HBV Modelled Discharge GRDC Station Kerki", ylabel="Discharge (m3/s)", xlabel="Time")
```
