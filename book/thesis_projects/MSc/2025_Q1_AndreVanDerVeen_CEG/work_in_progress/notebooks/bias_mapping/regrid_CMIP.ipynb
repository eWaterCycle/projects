```python
for file_name in [test_other_cmip_regridded.temperatureNC, test_other_cmip_regridded.precipitationNC]:
    dataset_regrid = xr.load_dataset(f"{test_other_cmip_regridded.directory}/{file_name}")
    print(dataset_regrid)
    print("------------------------")
    var = list(dataset_regrid.data_vars.keys())[0]
    dataset_regrid[var].isel(time=-1).plot(cmap="coolwarm", robust=True, size=5)
```

```python

n_cells = dataset.sizes["lat"] * dataset.sizes["lon"]

n_cells_regrid = dataset_regrid.sizes["lat"] * dataset_regrid.sizes["lon"]


print("Total grid cells:", n_cells)

print("Total grid cells:", n_cells_regrid)
```

```python
var = list(dataset.data_vars)[0]

mean_before = dataset[var].mean().item()
mean_after  = dataset_regrid[var].mean().item()

print("Mean precip before :", mean_before)
print("Mean precip after  :", mean_after)
print("Relative change (%) :", 
      100*(mean_after-mean_before)/mean_before)
```

```python
ts_before = dataset[var].mean(dim=["lat","lon"])
ts_after  = dataset_regrid[var].mean(dim=["lat","lon"])

(ts_before - ts_after).plot()
```

```python
yearly_before = dataset[var].resample(time="1YE").sum()*1000
yearly_after  = dataset_regrid[var].resample(time="1YE").sum()*1000

yearly_before_mean = yearly_before.mean(dim=["lat","lon"])
yearly_after_mean  = yearly_after.mean(dim=["lat","lon"])
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

yearly_before_mean.plot(label="CMIP before regrid")
yearly_after_mean.plot(label="CMIP after regrid")

plt.ylabel("Yearly precipitation")
plt.title("Yearly total precipitation comparison")
plt.grid(alpha=0.3, linestyle="--")
plt.legend()
plt.ylim(0, None)
plt.show()
```

```python
mean_before = yearly_before_mean.mean().item()
mean_after  = yearly_after_mean.mean().item()

print("Mean yearly precip before :", mean_before)
print("Mean yearly precip after  :", mean_after)
print(
    "Relative change (%) :",
    100 * (mean_after - mean_before) / mean_before
)
```

```python
import numpy as np

weights_before = np.cos(np.deg2rad(dataset.lat))
weights_after  = np.cos(np.deg2rad(dataset_regrid.lat))
```

```python
yearly_before_w = (
    dataset[var]
    .weighted(weights_before)
    .mean(("lat", "lon"))
    .resample(time="1YE")
    .sum()
    * 1000
)

yearly_after_w = (
    dataset_regrid[var]
    .weighted(weights_after)
    .mean(("lat", "lon"))
    .resample(time="1YE")
    .sum()
    * 1000
)
```

```python
print("Mean yearly precip before :", yearly_before_w.mean().item())
print("Mean yearly precip after  :", yearly_after_w.mean().item())

print(
    "Relative change (%) :",
    100 * (yearly_after_w.mean() - yearly_before_w.mean()) / yearly_before_w.mean()
)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(6,6))

yearly_before_w.plot(label="Before regrid")
yearly_after_w.plot(label="After regrid")

plt.ylabel("Yearly precipitation (mm)")
plt.title("Yearly precipitation comparison")
plt.grid(alpha=0.3, linestyle="--")
plt.legend()
plt.savefig("regrid_yearly_comp.pdf", bbox_inches="tight")
plt.show()
```

```python
plt.figure(figsize=(6,6))

plt.scatter(yearly_before_w, yearly_after_w)

min_val = min(yearly_before_w.min(), yearly_after_w.min())
max_val = max(yearly_before_w.max(), yearly_after_w.max())

plt.plot([min_val, max_val], [min_val, max_val], 'k--')  # 1:1 line

plt.xlabel("Before regridding (mm/year)")
plt.ylabel("After regridding (mm/year)")
plt.title("Precipitation conservation check")

plt.grid(alpha=0.3, linestyle="--")
plt.savefig("regrid_conservation.pdf", bbox_inches="tight")
plt.show()
```

```python
plt.figure(figsize=(6,6))

(yearly_after_w - yearly_before_w).plot()

plt.axhline(0, linestyle="--")
plt.ylabel("Difference (mm/year)")
plt.title("Regridding-induced precipitation difference")
plt.grid(alpha=0.3, linestyle="--")
plt.savefig("regrid_diff.pdf", bbox_inches="tight")
plt.show()
```
