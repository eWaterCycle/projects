```python
from cmethods import adjust


def run_bias_corrections(obs, simh, simp, n_quantiles=250, group_monthly="time.month"):
    """
    Run a standard suite of bias-correction methods using cmethods.adjust.

    Parameters
    ----------
    obs : xarray.DataArray
        Observations (historical)
    simh : xarray.DataArray
        Historical simulation
    simp : xarray.DataArray
        Future simulation
    n_quantiles : int
        Number of quantiles for QM / QDM
    group_monthly : str
        Grouping for seasonal methods (e.g. 'time.month')

    Returns
    -------
    dict[str, xarray.DataArray]
        Dictionary with bias-corrected future simulations
    """

    results = {}

    results["qdm"] = adjust(
        method="quantile_delta_mapping",
        obs=obs,
        simh=simh,
        simp=simp,
        n_quantiles=n_quantiles,
        kind="*",
    )

    results["linear_scaling"] = adjust(
        method="linear_scaling", obs=obs, simh=simh, simp=simp, kind="*", group=group_monthly
    )

    results["variance_scaling"] = adjust(
        method="variance_scaling", obs=obs, simh=simh, simp=simp, kind="+", group=group_monthly
    )

    results["delta_scaling"] = adjust(
        method="delta_method", obs=obs, simh=simh, simp=simp, kind="*", group=group_monthly
    )

    results["qm"] = adjust(
        method="quantile_mapping", obs=obs, simh=simh, simp=simp, n_quantiles=n_quantiles, kind="*"
    )

    return results
```

```python
df_d = df.copy()  # (dropna?)

# convert daily aligned pandas to xarray DataArrays with explicit names
ds_obsh = xr.Dataset(
    data_vars={"discharge": ("time", df_d["Observed"].values)}, coords={"time": df_d.index}
)

ds_simh = xr.Dataset(
    data_vars={"discharge": ("time", df_d["CMIP_historical"].values)}, coords={"time": df_d.index}
)

ds_simp = xr.Dataset(
    data_vars={"discharge": ("time", df_d["CMIP_future"].values)}, coords={"time": df_d.index}
)
```

```python
bc = run_bias_corrections(
    obs=ds_obsh["discharge"], simh=ds_simh["discharge"], simp=ds_simp["discharge"]
)

qdm_adjusted = bc["qdm"]
linear_scaling = bc["linear_scaling"]
variance_scaling = bc["variance_scaling"]
delta_scaling = bc["delta_scaling"]
qm_adjusted = bc["qm"]
```

```python
def plot_year_xr_2(ds, label, year, var="discharge", linewidth=1.5, year_offset=0):
    """
    Plot a single year from an xarray Dataset, with optional year shift.

    Parameters
    ----------
    ds : xarray.Dataset
        Dataset containing the variable.
    label : str
        Label for the line.
    year : int
        Year to select (e.g., 1960 or 2060).
    var : str
        Variable name in the dataset.
    linewidth : float
        Line width for the plot.
    year_offset : int
        Number of years to shift the x-axis (positive or negative).
    """
    da_y = ds[var].sel(time=str(year))

    # Shift time if requested
    if year_offset != 0:
        shifted_time = da_y.time.to_index() + pd.DateOffset(years=year_offset)
    else:
        shifted_time = da_y.time.to_index()

    plt.plot(shifted_time, da_y, label=label, alpha=0.8, linewidth=linewidth)
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import xarray as xr


def plot_year_xr(ds, label, year, var=None, linewidth=1.5, year_offset=0):
    """
    Plot a single year from xarray Dataset or DataArray, with optional year shift.

    Parameters
    ----------
    ds : xarray.Dataset or DataArray
        Dataset or DataArray containing the variable
    label : str
        Line label
    year : int
        Year to plot
    var : str
        Variable name (required if ds is Dataset)
    linewidth : float
        Line width
    year_offset : int
        Years to shift the x-axis (positive or negative)
    """
    # Extract DataArray if Dataset
    if isinstance(ds, xr.Dataset):
        if var is None:
            # automatically pick first variable
            var = list(ds.data_vars)[0]
        da = ds[var].sel(time=str(year))
    else:
        da = ds.sel(time=str(year))

    # Shift time if requested
    if year_offset != 0:
        shifted_time = da.time.to_index() + pd.DateOffset(years=year_offset)
    else:
        shifted_time = da.time.to_index()

    plt.plot(shifted_time, da, label=label, alpha=0.8, linewidth=linewidth)
```

```python
plt.figure(figsize=(12, 5))

plot_year_xr(ds_obsh, "Observations", 1965, linewidth=3)
plot_year_xr(ds_simh, "HBV with CMIP_HIST", 1965, linewidth=3)
plot_year_xr(ds_simp, "HBV with CMIP_FUT (Y=2065, ssp585)", 2065, linewidth=3, year_offset=-100)

# BC methods
plot_year_xr(bc["qdm"], "CMIP_FUT Bias corrected: QDM", 2065, var="discharge", year_offset=-100)
plot_year_xr(
    bc["linear_scaling"],
    "CMIP_FUT Bias corrected: linear_scaling",
    2065,
    var="discharge",
    year_offset=-100,
)
plot_year_xr(
    bc["variance_scaling"],
    "CMIP_FUT Bias corrected: variance_scaling",
    2065,
    var="discharge",
    year_offset=-100,
)
plot_year_xr(
    bc["delta_scaling"],
    "CMIP_FUT Bias corrected: delta_scaling",
    1965,
    var="discharge",
    year_offset=0,
)
plot_year_xr(bc["qm"], "CMIP_FUT Bias corrected: qm", 2065, var="discharge", year_offset=-100)

plt.grid(linestyle="--", alpha=0.5)
plt.title("Test bias correction - Discharge @ GRDC Kerki")
plt.ylabel("Discharge [m³/s]")
plt.legend()
plt.show()
```
