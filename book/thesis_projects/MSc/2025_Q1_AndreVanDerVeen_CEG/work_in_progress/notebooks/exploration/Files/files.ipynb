```python
import netCDF4 as nc

file_path = "/data/shared/parameter-sets/pcrglobwb_global/global_30min/meteo/forcing/daily_temperature_cru_era-interim_2000_to_2010.nc4"

# Open the NetCDF file
ds = nc.Dataset(file_path)

# Print dimensions
print("Dimensions:")
for name, dim in ds.dimensions.items():
    print(f"  {name}: {len(dim)}")

# Print variables
print("\nVariables:")
for name, var in ds.variables.items():
    print(f"  {name}: shape={var.shape}, dimensions={var.dimensions}")

# Print global attributes
print("\nGlobal attributes:")
for attr in ds.ncattrs():
    print(f"  {attr} = {getattr(ds, attr)}")

ds.close()
```
