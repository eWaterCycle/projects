```python
# Template: allocate agriculture demand to nearest filtered source pixels.
# Required demand_df columns: ['lat', 'lon', 'demand']

import pandas as pd
import numpy as np


def haversine_km(lat1, lon1, lat2, lon2):
    r = 6371.0
    lat1r = np.radians(lat1)
    lon1r = np.radians(lon1)
    lat2r = np.radians(lat2)
    lon2r = np.radians(lon2)
    dlat = lat2r - lat1r
    dlon = lon2r - lon1r
    a = np.sin(dlat / 2.0) ** 2 + np.cos(lat1r) * np.cos(lat2r) * np.sin(dlon / 2.0) ** 2
    return 2.0 * r * np.arcsin(np.sqrt(a))


def allocate_water(demand_df, source_df, k_neighbors=3, max_distance_km=75.0):
    src = source_df.copy().reset_index(drop=True)
    dem = demand_df.copy().reset_index(drop=True)

    allocations = []
    unmet = []

    for d_idx, d in dem.iterrows():
        demand_left = float(d["demand"])

        valid = src["available"] > 0
        if not valid.any() or demand_left <= 0:
            unmet.append({"demand_id": d_idx, "unmet_demand": demand_left})
            continue

        src_valid = src.loc[valid]
        distances = haversine_km(
            d["lat"],
            d["lon"],
            src_valid["lat"].to_numpy(),
            src_valid["lon"].to_numpy(),
        )

        candidates = src_valid.assign(distance_km=distances)
        candidates = candidates[candidates["distance_km"] <= max_distance_km]
        candidates = candidates.nsmallest(k_neighbors, "distance_km")

        if candidates.empty:
            unmet.append({"demand_id": d_idx, "unmet_demand": demand_left})
            continue

        for s_idx, s in candidates.iterrows():
            if demand_left <= 0:
                break
            available = float(src.at[s_idx, "available"])
            if available <= 0:
                continue

            take = min(demand_left, available)
            src.at[s_idx, "available"] = available - take
            demand_left -= take

            allocations.append(
                {
                    "demand_id": d_idx,
                    "source_id": s_idx,
                    "allocated": take,
                    "distance_km": float(s["distance_km"]),
                    "demand_lat": float(d["lat"]),
                    "demand_lon": float(d["lon"]),
                    "source_lat": float(s["lat"]),
                    "source_lon": float(s["lon"]),
                }
            )

        if demand_left > 0:
            unmet.append({"demand_id": d_idx, "unmet_demand": demand_left})

    return pd.DataFrame(allocations), pd.DataFrame(unmet), src


# Build source table from your filtered coordinates.
withdrawal_fraction = 0.2  # tune this rule
source_df = coords_df[["lat", "lon", var_name]].copy()
source_df = source_df.rename(columns={var_name: "source_value"})
source_df["available"] = source_df["source_value"] * withdrawal_fraction


if "demand_df" not in globals():
    print("Define demand_df first with columns ['lat', 'lon', 'demand'] and run this cell again.")
else:
    allocation_df, unmet_df, source_after_df = allocate_water(
        demand_df,
        source_df,
        k_neighbors=3,
        max_distance_km=75.0,
    )
    print(f"Allocations: {len(allocation_df)} rows")
    print(f"Unmet demand pixels: {len(unmet_df)}")
    display(allocation_df.head(10))
    display(unmet_df.head(10))
```

```python
# Precompute nearest source pixel for every (lat, lon) demand grid point.
import numpy as np
import pandas as pd

if "dummy_water_demand" not in water_demand:
    raise ValueError("dummy_water_demand not found. Run cell 6 first.")
if "channel_storage" not in dummy_channel_storage.columns:
    raise ValueError("dummy_channel_storage['channel_storage'] not found. Run cell 14 first.")

# Unique demand grid coordinates (independent of timestep values).
base_demand_grid_df = (
    water_demand["dummy_water_demand"]
    .isel(time=0)
    .to_series()
    .reset_index(name="tmp")[["lat", "lon"]]
    .drop_duplicates()
    .reset_index(drop=True)
)

src_map_df = dummy_channel_storage[["lat", "lon"]].reset_index(drop=True)
src_lat = src_map_df["lat"].to_numpy()
src_lon = src_map_df["lon"].to_numpy()

nearest_ids = []
for _, row in base_demand_grid_df.iterrows():
    d_lat = float(row["lat"])
    d_lon = float(row["lon"])
    dist2 = (src_lat - d_lat) ** 2 + (src_lon - d_lon) ** 2
    nearest_ids.append(int(np.argmin(dist2)))

nearest_source_map_df = base_demand_grid_df.copy()
nearest_source_map_df["source_id"] = nearest_ids
nearest_source_map_df["source_lat"] = src_map_df.loc[nearest_source_map_df["source_id"], "lat"].to_numpy()
nearest_source_map_df["source_lon"] = src_map_df.loc[nearest_source_map_df["source_id"], "lon"].to_numpy()

print(f"Nearest-source map built for {len(nearest_source_map_df)} demand grid pixels")
nearest_source_map_df.head(10)
```

```python
# Simulate one model step using precomputed nearest-source map.
import numpy as np
import pandas as pd

rng = np.random.default_rng(42)

if "dummy_water_demand" not in water_demand:
    raise ValueError("dummy_water_demand not found. Run cell 6 first.")
if "channel_storage" not in dummy_channel_storage.columns:
    raise ValueError("dummy_channel_storage['channel_storage'] not found. Run cell 14 first.")
if "nearest_source_map_df" not in globals():
    raise ValueError("nearest_source_map_df not found. Run cell 17 first.")

# 1) Pick a random timestep from dummy demand.
time_name = "time" if "time" in water_demand["dummy_water_demand"].dims else water_demand["dummy_water_demand"].dims[0]
nt = water_demand["dummy_water_demand"].sizes[time_name]
random_t_idx = int(rng.integers(0, nt))

wd_step = water_demand["dummy_water_demand"].isel({time_name: random_t_idx})
wd_df = wd_step.to_series().dropna().reset_index(name="demand")
wd_df = wd_df[wd_df["demand"] > 0].reset_index(drop=True)

# Optional cap for faster experimentation.
max_demand_points = 5000
if len(wd_df) > max_demand_points:
    wd_df = wd_df.sample(n=max_demand_points, random_state=42).reset_index(drop=True)

# 2) Join with precomputed nearest source ids.
wd_df = wd_df.merge(nearest_source_map_df, on=["lat", "lon"], how="left")

# 3) Prepare source table.
src_df = dummy_channel_storage[["lat", "lon", "channel_storage"]].copy().reset_index(drop=True)
src_df["remaining_storage"] = src_df["channel_storage"].astype(float)

# 4) Extract demand from mapped source.
alloc_rows = []
unmet_rows = []

for d_idx, d in wd_df.iterrows():
    d_need = float(d["demand"])
    nearest_src = int(d["source_id"])

    available = float(src_df.at[nearest_src, "remaining_storage"])
    taken = min(d_need, max(0.0, available))
    src_df.at[nearest_src, "remaining_storage"] = available - taken
    unmet = d_need - taken

    alloc_rows.append(
        {
            "demand_id": d_idx,
            "demand_lat": float(d["lat"]),
            "demand_lon": float(d["lon"]),
            "demand": d_need,
            "source_id": nearest_src,
            "source_lat": float(d["source_lat"]),
            "source_lon": float(d["source_lon"]),
            "extracted": taken,
            "unmet": unmet,
        }
    )

    if unmet > 0:
        unmet_rows.append({"demand_id": d_idx, "unmet": unmet})

allocation_result_df = pd.DataFrame(alloc_rows)
unmet_result_df = pd.DataFrame(unmet_rows)
source_after_step_df = src_df

print(f"Random timestep index: {random_t_idx} ({time_name})")
print(f"Demand pixels used: {len(wd_df)}")
print(f"Total demand: {allocation_result_df['demand'].sum():.3f}")
print(f"Total extracted: {allocation_result_df['extracted'].sum():.3f}")
print(f"Total unmet: {allocation_result_df['unmet'].sum():.3f}")
print(f"Source pixels depleted: {(source_after_step_df['remaining_storage'] <= 0).sum()}")

allocation_result_df.head(10)
```

```python
# Template function: allocate extraction for one timestep using a precomputed nearest-source map.
import numpy as np
import pandas as pd
import xarray as xr


def allocate_extraction_timestep(
    demand_t_da: xr.DataArray,
    storage_t_da: xr.DataArray,
    nearest_source_map_df: pd.DataFrame,
    max_demand_points: int | None = None,
):
    # demand_t_da: 2D DataArray with dims (lat, lon) and demand values
    # storage_t_da: 2D DataArray with dims (lat, lon) and channel storage values
    # nearest_source_map_df columns: lat, lon, source_id, source_lat, source_lon

    wd_df = demand_t_da.to_series().dropna().reset_index(name="demand")
    wd_df = wd_df[wd_df["demand"] > 0].reset_index(drop=True)

    if max_demand_points is not None and len(wd_df) > max_demand_points:
        wd_df = wd_df.sample(n=max_demand_points, random_state=42).reset_index(drop=True)

    wd_df = wd_df.merge(nearest_source_map_df, on=["lat", "lon"], how="left")

    src_df = storage_t_da.to_series().dropna().reset_index(name="remaining_storage")
    src_df = src_df.rename(columns={"lat": "source_lat", "lon": "source_lon"}).reset_index(drop=True)

    alloc_rows = []
    unmet_rows = []

    for d_idx, d in wd_df.iterrows():
        s_idx = int(d["source_id"])
        d_need = float(d["demand"])

        available = float(src_df.at[s_idx, "remaining_storage"])
        taken = min(d_need, max(0.0, available))
        src_df.at[s_idx, "remaining_storage"] = available - taken
        unmet = d_need - taken

        alloc_rows.append(
            {
                "demand_id": d_idx,
                "lat": float(d["lat"]),
                "lon": float(d["lon"]),
                "demand": d_need,
                "source_id": s_idx,
                "source_lat": float(d["source_lat"]),
                "source_lon": float(d["source_lon"]),
                "extracted": taken,
                "unmet": unmet,
            }
        )

        if unmet > 0:
            unmet_rows.append({"demand_id": d_idx, "unmet": unmet})

    allocation_df = pd.DataFrame(alloc_rows)
    unmet_df = pd.DataFrame(unmet_rows)

    # Rebuild updated storage DataArray with original shape/coords.
    updated_storage_t_da = storage_t_da.copy()
    updated_series = src_df.set_index(["source_lat", "source_lon"])["remaining_storage"]
    updated_storage_t_da.values = (
        updated_series
        .reindex(storage_t_da.to_series().index)
        .to_numpy()
        .reshape(storage_t_da.shape)
    )

    return updated_storage_t_da, allocation_df, unmet_df
```
