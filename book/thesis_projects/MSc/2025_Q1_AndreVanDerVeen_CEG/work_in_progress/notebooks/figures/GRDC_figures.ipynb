# Images for report - GRDC

```python
import matplotlib.pyplot as plt
from ewatercycle.observation.grdc import get_grdc_data
```

```python
grdc_chatly = get_grdc_data(
    2817100, "1900-01-01T00:00Z", "2001-01-01T00:00Z", data_home="../../data/GRDC/Daily"
)

grdc_kerki = get_grdc_data(
    2617110, "1900-01-01T00:00Z", "2001-01-01T00:00Z", data_home="../../data/GRDC/Daily"
)
```

```python
plt.figure(figsize=(5, 4))
grdc_kerki["streamflow"].plot(label="Kerki (upstream)", color="tab:orange", linewidth=0.75)
grdc_chatly["streamflow"].plot(label="Chatly (downstream)", color="tab:blue", linewidth=0.75)

plt.title("GRDC Discharge Data for Chatly and Kerki Stations")
plt.grid(True, linestyle="--", alpha=0.5)
plt.legend()
```

```python
plt.figure(figsize=(5, 4))

grdc_kerki.sel(time=slice("1960-01-01", "1960-12-31"))["streamflow"].plot(
    label="Kerki (upstream)", color="tab:orange"
)
grdc_chatly.sel(time=slice("1960-01-01", "1960-12-31"))["streamflow"].plot(
    label="Chatly (downstream)", color="tab:blue"
)


plt.title("Streamflow in 1960 at Chatly and Kerki (GRDC Stations)")
plt.xlabel("Date")
plt.ylabel("Streamflow [m³/s]")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
fig, ax = plt.subplots(1, 2, figsize=(10, 4), sharey=False)

# --- LEFT: Full time series ---
grdc_kerki["streamflow"].plot(ax=ax[0], label="Kerki (upstream)", color="tab:orange")
grdc_chatly["streamflow"].plot(ax=ax[0], label="Chatly (downstream)", color="tab:blue")

ax[0].set_title("Full Streamflow Record")
ax[0].set_xlabel("Date")
# ax[0].set_ylabel('Streamflow [m³/s]')
ax[0].legend()
ax[0].grid(True, linestyle="--", alpha=0.5)

# --- RIGHT: Only 1960 ---
grdc_kerki.sel(time=slice("1960-01-01", "1960-12-31"))["streamflow"].plot(
    ax=ax[1], label="Kerki (upstream)", color="tab:orange"
)
grdc_chatly.sel(time=slice("1960-01-01", "1960-12-31"))["streamflow"].plot(
    ax=ax[1], label="Chatly (downstream)", color="tab:blue"
)

ax[1].set_title("Streamflow in 1960")
ax[1].set_xlabel("Date")
ax[1].legend()
ax[1].grid(True, linestyle="--", alpha=0.5)

plt.suptitle(
    "Daily Average Streamflow at GRDC station Kerki (upstream) and GRDC station Chatly (downstream)",
    fontsize=12,
)
plt.tight_layout()
plt.savefig(
    "../../outputs/figures/GRDC_Streamflow_Chatly_Kerki.png",
    bbox_inches="tight",
    pad_inches=0.05,
    dpi=300,
)
plt.show()
```
