testing some optimization of shapefile calls

```python
# setup imports
import sys
from pathlib import Path

import numpy as np

PROJECT_ROOT = Path().resolve().parents[2]  # pas aan als notebook dieper/dichter zit
sys.path.append(str(PROJECT_ROOT))
```

```python
import time

import pandas as pd

from src.utils import catchment_area_from_shapefile, mmday_to_m3s

# Fake model output
model_output = pd.Series(np.random.rand(5 * 365))  # 1000 timesteps
shape_name = "Kerki_GRDC"


# --- Old method ---
def mmday_old(model_output, shape_name):
    # compute area every time
    area = catchment_area_from_shapefile(shape_name)
    return model_output * (1e-3 / 86400 * area)


# --- New method ---
def mmday_new(model_output, shape_name):
    return mmday_to_m3s(model_output, shape_name)  # uses cached area


# Number of calls
n_calls = 720

# Time old
start = time.perf_counter()
for _ in range(n_calls):
    mmday_old(model_output, shape_name)
end = time.perf_counter()
print("Old method:", end - start, "seconds")

# Time new
start = time.perf_counter()
for _ in range(n_calls):
    mmday_new(model_output, shape_name)
end = time.perf_counter()
print("New method:", end - start, "seconds")
```
