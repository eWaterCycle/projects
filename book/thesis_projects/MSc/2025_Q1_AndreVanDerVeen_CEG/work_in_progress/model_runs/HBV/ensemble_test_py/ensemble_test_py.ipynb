```python
# setup imports
import sys
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np

PROJECT_ROOT = Path().resolve().parents[2]  # pas aan als notebook dieper/dichter zit
sys.path.append(str(PROJECT_ROOT))

from ewatercycle.observation.grdc import get_grdc_data

from src.forcing import load_lumped_forcing
from src.models import run_ensemble_HBV
from src.paths import GRDC
from src.utils import catchment_area_from_shapefile
```

```python
ERA5_forcing_loaded = load_lumped_forcing("Kerki_GRDC", "ERA5", "1959-1963")
# plot_ERA5_forcing(ERA5_forcing_loaded)
```

```python
results = run_ensemble_HBV(n_particles=100, forcing=ERA5_forcing_loaded)
```

```python
plt.figure(figsize=(15, 5))
plt.plot(results, linewidth=1, alpha=0.1, color="black");
```

```python
area = catchment_area_from_shapefile("Kerki_GRDC")

results_m3s = results * area * 1e-3 / 86400
```

```python
grdc_station = get_grdc_data(
    2617110, "1955-01-01T00:00Z", "1963-12-31T00:00Z", data_home=GRDC / "Daily"
)
```

```python
start, end = "1959-01-01", "1963-12-31"

plt.figure(figsize=(15, 4))

grdc_station["streamflow"].sel(time=slice(start, end)).plot(
    label="GRDC", color="tab:orange", linewidth=3, zorder=3
)

plt.plot(results_m3s.loc[start:end], linewidth=1, alpha=0.1, color="black", zorder=2)

plt.ylim(0, np.max(grdc_station["streamflow"]))
plt.legend()
```
